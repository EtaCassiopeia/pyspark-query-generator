# 01 — Explore Training Data

Load and inspect the training and evaluation datasets.

```python
import sys
sys.path.insert(0, "..")

from shared.data_validator import validate_jsonl
from pathlib import Path
```

```python
# Load training data
train_examples = validate_jsonl(Path("../data/samples/training_data.jsonl"))
print(f"Training examples: {len(train_examples)}")

eval_examples = validate_jsonl(Path("../data/samples/evaluation_data.jsonl"))
print(f"Evaluation examples: {len(eval_examples)}")
```

```python
# Complexity distribution
from collections import Counter

complexity_dist = Counter(ex.complexity for ex in train_examples)
print("Complexity distribution:")
for level, count in complexity_dist.most_common():
    print(f"  {level}: {count}")
```

```python
# Most common operations
all_ops = [op for ex in train_examples for op in ex.operations]
op_dist = Counter(all_ops)
print("Top 10 operations:")
for op, count in op_dist.most_common(10):
    print(f"  {op}: {count}")
```

```python
# All referenced tables
all_tables = set(t for ex in train_examples for t in ex.tables)
print(f"Unique tables: {len(all_tables)}")
for t in sorted(all_tables):
    print(f"  {t}")
```

```python
# Inspect a single example
ex = train_examples[0]
print(f"Description: {ex.description}")
print(f"Tables: {ex.tables}")
print(f"Operations: {ex.operations}")
print(f"Complexity: {ex.complexity}")
print(f"\nCode:\n{ex.pyspark_code}")
```
